```python
import numpy as np

n_tot=17
B_c=1
for i in range(1,n_tot+1):
    B_c *= (365-i+1)

B_c *= 1/365**n_tot

print(1-B_c)
```

```python
rng = np.random.default_rng()

def pick_doors(n_doors=3):
    return rng.integers(3), rng.integers(3)

def host_pick(n_doors=3):
    con, car=pick_doors(n_doors)
    host=rng.integers(3)

    while host == con or host == car:
        host=rng.integers(3)

    return con, car, host
```

```python

n_att=10**5

n_switch=0
n_stick=0
```

```python
for i in range(0,n_att):
    con, car, host = host_pick(3)
    for j in range(0,3):
        if j != con and j != host:
            if j == car:
                n_switch += 1
            else:
                n_stick+=1

P_switch = n_switch/n_att
P_stick = n_stick/n_att

print('P_switch = ', P_switch)
print('P_stick = ', P_stick)
```
